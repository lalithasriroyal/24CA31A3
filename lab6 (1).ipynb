Consider the following sentences:

###  Answer the following questions: 

 ### 1. Which words have similar meanings?  
A.  Similar words:

love ↔ enjoy (similar meaning)

machine learning ↔ artificial intelligence (related concepts)

cats ↔ dogs (both animals)


### 2. How can Word2Vec represent these words numerically?  
A. Numerical representation in Word2Vec:

Each word is mapped to a dense vector (e.g., [0.25, -0.13, 0.89, ...]) that captures semantic meaning.

Words with similar meanings have vectors close together in the vector space.


### 3. Which architecture (CBOW or Skip-Gram) predicts the missing word?
A. Architecture predicting missing word:

CBOW (Continuous Bag of Words) → predicts the center word using surrounding context words.

### 4. Which architecture predicts the surrounding words? 
A. Architecture predicting surrounding words:

Skip‑Gram → uses the target word to predict its neighboring context words.

### IN LAb

### Algorithm :
### Step 1: Import the required libraries. 
### Step 2: Create a small text corpus. 
### Step 3: Preprocess the text by converting it into tokenized words. 
### Step 4: Train the Word2Vec model using the corpus. 
### Step 5: Generate vector representations for each word. 
### Step 6: Convert each sentence into a sentence embedding by averaging its word vectors. 
### Step 7: Compute the cosine similarity between the two sentence vectors. 
### Step 8: Display the similarity score.  
### Sample Input Sentence 
### 1: I love machine learning Sentence 
### 2: I enjoy artificial intelligence 
### Expected Output Sentence 
### 1 : I love machine learning Sentence 
### 2 : I enjoy artificial intelligence 
### Cosine Similarity : 0.82 

```python
# Step 1: Import libraries
from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

# Step 2: Create corpus
corpus = [
    "I love machine learning",
    "I enjoy artificial intelligence",
    "Cats drink milk",
    "Dogs drink milk"
]

# Step 3: Preprocess (tokenize)
tokenized_corpus = [sentence.lower().split() for sentence in corpus]

# Step 4: Train Word2Vec model
model = Word2Vec(sentences=tokenized_corpus, vector_size=50, window=3, min_count=1, sg=0)  # sg=0 → CBOW, sg=1 → Skip-Gram

# Step 5: Generate word vectors
print(model.wv['love'])
print(model.wv['enjoy'])

# Step 6: Sentence embeddings (average word vectors)
def sentence_vector(sentence):
    words = sentence.lower().split()
    return np.mean([model.wv[w] for w in words], axis=0)

vec1 = sentence_vector("I love machine learning")
vec2 = sentence_vector("I enjoy artificial intelligence")

# Step 7: Cosine similarity
similarity = cosine_similarity([vec1], [vec2])[0][0]

# Step 8: Display result
print("Sentence 1: I love machine learning")
print("Sentence 2: I enjoy artificial intelligence")
print("Cosine Similarity:", round(similarity, 2))
```
