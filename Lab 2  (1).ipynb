## Input Sentence

```python
text = "Dr. Ravi is studying Machine Learning at Anna University in Chennai."
```

## Tokenization

```python
tokens = word_tokenize(text)
```

## Stop-word & Punctuation Removal

```python
stop_words = set(stopwords.words("english"))

filtered_tokens = [
    word.replace(".", "")
    for word in tokens
    if word.lower() not in stop_words
    and word not in string.punctuation
]
```

## Lemmatization

```python
lemmatizer = WordNetLemmatizer()

lemmatized_tokens = []

for word in filtered_tokens:

    if word.lower() == "studying":
        lemma = lemmatizer.lemmatize(word.lower(), pos="v")
    else:
        lemma = word

    lemmatized_tokens.append(lemma)
```

## POS Tagging

```python
import nltk
nltk.download('averaged_perceptron_tagger_eng')
```

```python
pos_tags = nltk.pos_tag(lemmatized_tokens)
```

## Input Sentences

```python
sentences = [

    "Dr. Ravi is studying Machine Learning at Anna University in Chennai.",

    "The students are learning Artificial Intelligence using Python programming.",

    "Microsoft announced new AI technologies during the conference in Hyderabad.",

    "Doctors are treating patients using advanced medical equipment in Apollo Hospital.",

    "Researchers developed intelligent robots for industrial automation in India."

]
```

## Complete Python Program

```python
# ============================================================
# POST LAB ACTIVITY
# NLP PREPROCESSING PIPELINE
# ============================================================

import nltk
import string
import spacy

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

# Download resources
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('averaged_perceptron_tagger')

# Load spaCy model
nlp = spacy.load("en_core_web_sm")

# Initialize objects
stop_words = set(stopwords.words("english"))
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# Input Sentences
sentences = [

    "Dr. Ravi is studying Machine Learning at Anna University in Chennai.",

    "The students are learning Artificial Intelligence using Python programming.",

    "Microsoft announced new AI technologies during the conference in Hyderabad.",

    "Doctors are treating patients using advanced medical equipment in Apollo Hospital.",

    "Researchers developed intelligent robots for industrial automation in India."

]

# Process Each Sentence
for i, text in enumerate(sentences, start=1):

    print("="*80)
    print(f"Sentence {i}")
    print("="*80)

    print("\nOriginal Sentence")
    print(text)

    # Tokenization
    tokens = word_tokenize(text)

    print("\nTokenization")
    print(tokens)

    # Stop-word Removal
    filtered_tokens = [

        word.replace(".", "")

        for word in tokens

        if word.lower() not in stop_words

        and word not in string.punctuation

    ]

    print("\nStop-word & Punctuation Removal")
    print(filtered_tokens)

    # Stemming
    stemmed = [

        stemmer.stem(word.lower())

        for word in filtered_tokens

    ]

    print("\nStemming")
    print(stemmed)

    # Lemmatization
    lemmatized = []

    for word in filtered_tokens:

        lemma = lemmatizer.lemmatize(word.lower(), pos='v')

        lemmatized.append(lemma)

    print("\nLemmatization")
    print(lemmatized)

    # POS Tagging
    pos_tags = nltk.pos_tag(lemmatized)

    print("\nPOS Tagging")
    print(pos_tags)

    # Named Entity Recognition
    doc = nlp(text)

    print("\nNamed Entity Recognition")

    for ent in doc.ents:
        print(ent.text, ":", ent.label_)

    print("\n")
```

## Observation

### Stemming removes prefixes and suffixes using predefined rules and may produce non-dictionary words (e.g., technolog, studi).

### Lemmatization returns meaningful dictionary words by considering the grammatical role (Part-of-Speech) of each word.

### Lemmatization generally provides more accurate and readable outputs than stemming, making it more suitable for advanced NLP applications.

```python
# ============================================================
# POST LAB ACTIVITY
# STEMMING vs LEMMATIZATION COMPARISON
# ============================================================

import nltk

from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer

# Download resources
nltk.download('wordnet')
nltk.download('omw-1.4')

# Initialize
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# Input Words
words = [
    "studying",
    "learning",
    "machines",
    "universities",
    "running",
    "better",
    "children",
    "wolves",
    "technologies",
    "developed"
]

print("="*70)
print("{:<18}{:<20}{:<20}".format("Original Word","Stemmed","Lemmatized"))
print("="*70)

for word in words:

    stem = stemmer.stem(word)

    lemma = lemmatizer.lemmatize(word, pos='v')

    print("{:<18}{:<20}{:<20}".format(word, stem, lemma))
```

```python
import pandas as pd

comparison = []

for word in words:
    comparison.append([
        word,
        stemmer.stem(word),
        lemmatizer.lemmatize(word, pos='v')
    ])

df = pd.DataFrame(
    comparison,
    columns=["Original Word", "Stemmed", "Lemmatized"]
)

print(df)
```
