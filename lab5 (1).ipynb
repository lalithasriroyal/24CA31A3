## 3.2 TF-IDF Architecture

TF-IDF converts text into numerical features.

### Term Frequency (TF)

For a term *t* in document *d*:

\[
TF(t,d)=\frac{\text{Number of occurrences of }t\text{ in }d}
{\text{Total number of terms in }d}
\]

### Inverse Document Frequency (IDF)

\[
IDF(t)=\log\left(\frac{N+1}{DF(t)+1}\right)+1
\]

where:

- \(N\) = number of documents
- \(DF(t)\) = number of documents containing term \(t\)

### TF-IDF

\[
TFIDF(t,d)=TF(t,d)\times IDF(t)
\]

A word that occurs frequently in one document but not in many documents receives a relatively high TF-IDF value.

### TF-IDF processing pipeline

```text
Documents
    ↓
Tokenization / vocabulary construction
    ↓
Term frequency calculation
    ↓
Document frequency calculation
    ↓
IDF calculation
    ↓
TF × IDF
    ↓
Sparse numerical feature matrix
```

## 3.3 Naïve Bayes Architecture for Spam Detection

Naïve Bayes is a probabilistic classifier based on Bayes' theorem.

\[
P(C|X)=\frac{P(X|C)P(C)}{P(X)}
\]

For a document containing multiple words, the classifier assumes conditional independence between features:

\[
P(C|x_1,x_2,\ldots,x_n)
\propto
P(C)\prod_{i=1}^{n}P(x_i|C)
\]

The class with the highest posterior probability is selected.

### Architecture

```text
TF-IDF Features
      │
      ▼
Training Documents
      │
      ▼
Estimate Class Probabilities
P(Spam), P(Not Spam)
      │
      ▼
Estimate Feature/Class Probabilities
      │
      ▼
Multinomial Naïve Bayes Model
      │
      ▼
New Email → TF-IDF
      │
      ▼
Posterior Probability for Each Class
      │
      ▼
Highest Probability
      │
      ▼
Spam / Not Spam
```

### Important characteristics

- Probabilistic classifier.
- Fast to train and predict.
- Effective for many text classification problems.
- Multinomial Naïve Bayes is particularly suitable for word-count/text features.
- `alpha` provides smoothing to avoid zero probabilities.

## 3.4 SVM Architecture for Sentiment Analysis

SVM attempts to find a decision boundary that separates classes with the largest possible margin.

For a linear SVM:

\[
f(x)=w^Tx+b
\]

The predicted class depends on the side of the decision boundary on which the feature vector lies.

### Architecture

```text
Customer Reviews
      │
      ▼
Text Preprocessing
      │
      ▼
TF-IDF Feature Matrix
      │
      ▼
Training Samples
      │
      ▼
Linear SVM
      │
      ├── Find separating hyperplane
      ├── Maximize margin
      └── Use support vectors
      │
      ▼
Trained SVM Model
      │
      ▼
New Review → TF-IDF
      │
      ▼
Decision Function
      │
      ▼
Positive / Negative / Neutral
```

### Important characteristics

- Effective for high-dimensional feature spaces.
- Works particularly well with sparse TF-IDF matrices.
- Linear SVM is computationally efficient for text classification.
- The `C` parameter controls the trade-off between margin width and training errors.
- With three sentiment classes, the classifier learns multiple class boundaries internally.

## 3.5 Model Evaluation Architecture

```text
Actual Class ──────────┐
                       ▼
                 ┌─────────────┐
Predicted Class ─► Evaluation  │
                 └──────┬──────┘
                        │
              ┌─────────┴─────────┐
              ▼                   ▼
          Accuracy          Confusion Matrix
```

Accuracy:

\[
Accuracy=\frac{Correct\ Predictions}{Total\ Predictions}
\]

For classification problems, the confusion matrix shows how many samples from each actual class were predicted as each class.

# 5. Activity 1 – Spam Detection Using Naïve Bayes


## Step 1: Create the Spam Email Dataset

The dataset contains two classes:

- **Spam**
- **Not Spam**

For demonstration, a larger small dataset is created so that the train/test split is meaningful. In a real application, hundreds or thousands of labelled emails would normally be used.


````python

spam_emails = [
    "Congratulations you won a lottery",
    "Claim your free prize now",
    "You have won a cash reward",
    "Free gift waiting for you",
    "Congratulations claim your bonus",
    "Win a free vacation today",
    "You are selected for a prize",
    "Claim your lottery reward",
    "Get your free cash prize",
    "Special offer claim now",
    "You won a free gift",
    "Congratulations you are a winner",
    "Earn money with this special offer",
    "Free coupon available claim today",
    "Limited time prize offer",
    "Win exciting rewards now",

    "Meeting at 10 AM tomorrow",
    "Project report submission is due",
    "Please attend the department meeting",
    "The class starts at 9 AM",
    "Submit the assignment by Friday",
    "Your project review is scheduled tomorrow",
    "Please send the meeting minutes",
    "The examination timetable is attached",
    "Faculty meeting is scheduled today",
    "Please complete the project report",
    "The seminar begins at 11 AM",
    "Submit your lab record tomorrow",
    "The project presentation is next week",
    "Please attend the academic review",
    "The department meeting is postponed",
    "Assignment submission deadline is Friday"
]

spam_labels = [
    "Spam"] * 16 + ["Not Spam"] * 16

spam_df = pd.DataFrame({
    "email": spam_emails,
    "label": spam_labels
})

spam_df
````

## Step 2: Inspect the Dataset

````python

print("Dataset shape:", spam_df.shape)
print("\nClass distribution:")
print(spam_df["label"].value_counts())

spam_df
````


## Step 3: Convert Emails into TF-IDF Features

`TfidfVectorizer` performs the conversion from raw text into a numerical matrix.

Important parameters used:

- `lowercase=True` – converts text to lowercase.
- `stop_words="english"` – removes common English stop words.
- `ngram_range=(1,2)` – considers unigrams and bigrams.


````python

X_text = spam_df["email"]
y = spam_df["label"]

tfidf_spam = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2)
)

X_tfidf = tfidf_spam.fit_transform(X_text)

print("TF-IDF matrix shape:", X_tfidf.shape)
print("Vocabulary size:", len(tfidf_spam.vocabulary_))
````

### Inspect the TF-IDF Vocabulary

````python

feature_names = tfidf_spam.get_feature_names_out()

print("First 30 TF-IDF features:")
print(feature_names[:30])

tfidf_preview = pd.DataFrame(
    X_tfidf.toarray(),
    columns=feature_names
)

tfidf_preview.head()
````


## Step 4: Split the Dataset

The dataset is divided into:

- **Training data** – used to learn the classification model.
- **Testing data** – used to evaluate the model on unseen emails.

`stratify=y` maintains approximately the same class distribution in both sets.


````python

X_train, X_test, y_train, y_test = train_test_split(
    X_tfidf,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("Training samples:", X_train.shape[0])
print("Testing samples :", X_test.shape[0])
````


## Step 5: Train the Multinomial Naïve Bayes Classifier


````python

nb_model = MultinomialNB(alpha=1.0)
nb_model.fit(X_train, y_train)

print("Naïve Bayes model trained successfully.")
````

## Step 6: Predict Test Emails

````python

y_pred_nb = nb_model.predict(X_test)

results_nb = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred_nb
})

results_nb
````

## Step 7: Calculate Accuracy and Classification Report

````python

accuracy_nb = accuracy_score(y_test, y_pred_nb)

print(f"Naïve Bayes Accuracy: {accuracy_nb:.2%}")
print("\nClassification Report:")
print(classification_report(y_test, y_pred_nb))
````

## Step 8: Display the Confusion Matrix

````python

labels_nb = ["Not Spam", "Spam"]

cm_nb = confusion_matrix(
    y_test,
    y_pred_nb,
    labels=labels_nb
)

plt.figure(figsize=(6, 4))
sns.heatmap(
    cm_nb,
    annot=True,
    fmt="d",
    xticklabels=labels_nb,
    yticklabels=labels_nb
)

plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.title("Naïve Bayes – Spam Detection Confusion Matrix")
plt.show()
````


### Understanding the Confusion Matrix

For the two-class problem:

- **True Negative (TN):** Not Spam correctly classified as Not Spam.
- **True Positive (TP):** Spam correctly classified as Spam.
- **False Positive (FP):** Not Spam incorrectly classified as Spam.
- **False Negative (FN):** Spam incorrectly classified as Not Spam.


## Step 9: Predict New Email Messages

````python

new_emails = [
    "Congratulations you have won a free prize",
    "Please submit your project report tomorrow",
    "Claim your cash reward now",
    "The department meeting is at 10 AM"
]

new_email_tfidf = tfidf_spam.transform(new_emails)
new_predictions = nb_model.predict(new_email_tfidf)

new_email_results = pd.DataFrame({
    "New Email": new_emails,
    "Predicted Class": new_predictions
})

new_email_results
````


# 6. Activity 2 – Sentiment Analysis Using SVM

## Step 1: Create the Customer Review Dataset

The three sentiment classes are:

- Positive
- Negative
- Neutral

The dataset below is deliberately small for laboratory demonstration. For reliable sentiment analysis, use a substantially larger labelled dataset.


````python

reviews = [
    "This phone is excellent",
    "The camera quality is amazing",
    "I love this product",
    "The battery life is fantastic",
    "Very good performance and excellent design",
    "This is a wonderful phone",
    "The product works perfectly",
    "I am very happy with this purchase",
    "Excellent quality and great performance",
    "The phone is fast and reliable",

    "Battery performance is poor",
    "The camera quality is terrible",
    "I hate this product",
    "The phone is very slow",
    "Very bad performance",
    "The product stopped working",
    "I am unhappy with this purchase",
    "The quality is disappointing",
    "Poor battery and poor camera",
    "This phone is awful",

    "The product is okay",
    "The phone is average",
    "The camera is acceptable",
    "Battery performance is normal",
    "The product is neither good nor bad",
    "It works as expected",
    "The design is standard",
    "The performance is moderate",
    "The phone is fine",
    "The product is satisfactory"
]

sentiments = (
    ["Positive"] * 10 +
    ["Negative"] * 10 +
    ["Neutral"] * 10
)

review_df = pd.DataFrame({
    "review": reviews,
    "sentiment": sentiments
})

review_df
````

## Step 2: Inspect the Dataset

````python

print("Dataset shape:", review_df.shape)
print("\nSentiment distribution:")
print(review_df["sentiment"].value_counts())

review_df
````


## Step 3: Convert Reviews into TF-IDF Features


````python

X_review_text = review_df["review"]
y_review = review_df["sentiment"]

tfidf_review = TfidfVectorizer(
    lowercase=True,
    stop_words="english",
    ngram_range=(1, 2)
)

X_review_tfidf = tfidf_review.fit_transform(X_review_text)

print("TF-IDF matrix shape:", X_review_tfidf.shape)
print("Vocabulary size:", len(tfidf_review.vocabulary_))
````

## Step 4: Split the Review Dataset

````python

X_train_svm, X_test_svm, y_train_svm, y_test_svm = train_test_split(
    X_review_tfidf,
    y_review,
    test_size=0.30,
    random_state=42,
    stratify=y_review
)

print("Training samples:", X_train_svm.shape[0])
print("Testing samples :", X_test_svm.shape[0])
````


## Step 5: Train the Linear SVM Classifier

`LinearSVC` is a linear Support Vector Machine implementation that is well suited to sparse TF-IDF text features.

The parameter `C` controls the regularization strength:

- Smaller `C` → stronger regularization.
- Larger `C` → weaker regularization and a greater emphasis on fitting the training data.


````python

svm_model = LinearSVC(C=1.0, random_state=42)
svm_model.fit(X_train_svm, y_train_svm)

print("SVM model trained successfully.")
````

## Step 6: Predict Test Reviews

````python

y_pred_svm = svm_model.predict(X_test_svm)

results_svm = pd.DataFrame({
    "Actual": y_test_svm.values,
    "Predicted": y_pred_svm
})

results_svm
````

## Step 7: Calculate Accuracy and Classification Report

````python

accuracy_svm = accuracy_score(y_test_svm, y_pred_svm)

print(f"SVM Accuracy: {accuracy_svm:.2%}")
print("\nClassification Report:")
print(classification_report(y_test_svm, y_pred_svm))
````

## Step 8: Display the Confusion Matrix

````python

labels_svm = ["Negative", "Neutral", "Positive"]

cm_svm = confusion_matrix(
    y_test_svm,
    y_pred_svm,
    labels=labels_svm
)

plt.figure(figsize=(6, 5))
sns.heatmap(
    cm_svm,
    annot=True,
    fmt="d",
    xticklabels=labels_svm,
    yticklabels=labels_svm
)

plt.xlabel("Predicted Sentiment")
plt.ylabel("Actual Sentiment")
plt.title("SVM – Sentiment Analysis Confusion Matrix")
plt.show()
````

## Step 9: Predict Sentiment of New Reviews

````python

new_reviews = [
    "The phone has excellent performance",
    "The battery is very poor",
    "The product is average and acceptable",
    "I am extremely happy with the phone",
    "The camera quality is terrible"
]

new_review_tfidf = tfidf_review.transform(new_reviews)
new_predictions_svm = svm_model.predict(new_review_tfidf)

new_review_results = pd.DataFrame({
    "New Review": new_reviews,
    "Predicted Sentiment": new_predictions_svm
})

new_review_results
````

## Detailed Data Flow

### Training phase

1. Collect labelled text documents.
2. Clean and normalize the text.
3. Learn the TF-IDF vocabulary from the training corpus.
4. Transform training documents into TF-IDF vectors.
5. Train the classifier using the labelled vectors.
6. Store the trained model and vectorizer.

### Testing phase

1. Take an unseen document.
2. Apply the same TF-IDF vectorizer learned during training.
3. Generate its numerical feature vector.
4. Pass the vector to the trained classifier.
5. Obtain the predicted class.
6. Compare the prediction with the true class.

> **Important:** In a production workflow, the TF-IDF vectorizer should be fitted only on training data and then used to transform validation/test/new data. This prevents information leakage.



# 9. Algorithm: Naïve Bayes – Step-by-Step

### Input

A set of labelled emails.

### Procedure

1. Read the labelled email dataset.
2. Separate email text and class labels.
3. Split the data into training and testing sets.
4. Fit the TF-IDF vectorizer using training text.
5. Transform training and testing text into TF-IDF vectors.
6. Initialize Multinomial Naïve Bayes.
7. Estimate class prior probabilities.
8. Estimate feature probabilities for each class.
9. Apply smoothing using `alpha`.
10. Predict the class of each test email.
11. Calculate accuracy.
12. Generate the confusion matrix.
13. Transform new emails using the already fitted TF-IDF vectorizer.
14. Predict their classes.

### Output

**Spam** or **Not Spam**.



# 10. Algorithm: SVM – Step-by-Step

### Input

A set of labelled customer reviews.

### Procedure

1. Read the labelled review dataset.
2. Separate review text and sentiment labels.
3. Split the data into training and testing sets.
4. Fit the TF-IDF vectorizer using training text.
5. Transform training and testing reviews into TF-IDF vectors.
6. Initialize a Linear SVM.
7. Learn a separating decision boundary from the training vectors.
8. Determine the decision boundary that provides a suitable maximum margin.
9. Use the learned model to classify test reviews.
10. Calculate accuracy.
11. Generate the confusion matrix.
12. Transform new reviews using the fitted TF-IDF vectorizer.
13. Predict Positive, Negative, or Neutral sentiment.

### Output

**Positive / Negative / Neutral**.



# 11. Important Practical Notes

## 11.1 Why TF-IDF?

Machine-learning classifiers require numerical input. Raw sentences cannot directly be supplied to conventional scikit-learn classifiers. TF-IDF represents each document as a numerical vector.

## 11.2 Why Naïve Bayes for Spam?

Spam filtering often contains highly informative words such as promotional terms, reward-related terms, and suspicious phrases. Naïve Bayes can learn probabilistic relationships between these features and spam/non-spam classes efficiently.

## 11.3 Why SVM for Sentiment Analysis?

TF-IDF creates a high-dimensional sparse feature space. Linear SVM is well suited to such feature spaces and can learn effective decision boundaries.

## 11.4 Why use the same TF-IDF approach for both activities?

Using the same feature representation makes it easier to understand that **feature extraction and classification are separate stages**. The classifier can be changed while the TF-IDF representation remains the same.

## 11.5 Small Dataset Limitation

The datasets in this notebook are educational examples. Accuracy from such a small dataset should not be interpreted as real-world model performance. Real applications require larger, diverse, representative datasets and appropriate validation.
