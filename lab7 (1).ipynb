### Exercise 7: Word Embeddings using GloVe 

### Aim 
To understand the concept of GloVe (Global Vectors for Word Representation) and use pre-trained GloVe embeddings to measure the semantic similarity between two sentences. 

### Objectives 

After completing the pre-lab, students will be able to: 
1. Understand the concept of word embeddings. .  
2. Understand the working principle of GloVe. 
3. Explain how GloVe uses global word co-occurrence statistics.  

### Pre Lab
Consider the following sentences: 
1. King and Queen rule the kingdom.  
2. Prince belongs to the royal family. 
3. Cat drinks milk.  
4. Dog drinks milk.  

### Answer the following questions: 

### 1. Which words have similar meanings?  
A.  Similar words:

King ↔ Queen ↔ Prince (royalty)

Cat ↔ Dog (animals)

### 2. Why do "King" and "Queen" have similar meanings?  
A. Both are royal titles and occur in similar contexts (kingdom, rule, royalty).

Their embeddings are close in vector space because of shared co‑occurrence patterns.

### 3. How does GloVe learn relationships between words?  
A. GloVe uses global word co‑occurrence statistics from a large corpus.

It builds a co‑occurrence matrix and factorizes it to learn embeddings where ratios of probabilities capture semantic meaning.

### 4. Why are pre‑trained embeddings useful?
A. Trained on huge corpora → capture rich semantic knowledge.

Save time and resources (no need to train from scratch).

Improve performance on NLP tasks with limited data.

### 5. Mention two applications of GloVe ?
A.  Sentiment analysis

Machine translation

Text classification

Semantic similarity tasks
