```python
import os
import urllib.request
import zipfile

url = "https://nlp.stanford.edu/data/glove.6B.zip"
zip_file = "glove.6B.zip"

if not os.path.exists("glove.6B.100d.txt"):
    print("Downloading GloVe model...")

    urllib.request.urlretrieve(url, zip_file)

    print("Download completed!")
    print("Extracting GloVe model...")

    with zipfile.ZipFile(zip_file, "r") as zip_ref:
        zip_ref.extract("glove.6B.100d.txt", ".")

    print("GloVe model extracted successfully!")

else:
    print("GloVe model already exists.")
```

```python
import numpy as np

glove_model = {}

print("Loading GloVe model...")

with open("glove.6B.100d.txt", "r", encoding="utf-8") as file:
    for line in file:
        values = line.split()

        word = values[0]
        vector = np.asarray(values[1:], dtype="float32")

        glove_model[word] = vector

print("GloVe model loaded successfully!")
print("Total words:", len(glove_model))
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# Function to create sentence embedding
def sentence_embedding(sentence):

    words = sentence.lower().split()

    word_vectors = []

    for word in words:
        if word in glove_model:
            word_vectors.append(glove_model[word])

    if len(word_vectors) == 0:
        return None

    # Average the word vectors
    sentence_vector = np.mean(word_vectors, axis=0)

    return sentence_vector


# Input sentences
sentence1 = input("Enter Sentence 1: ")
sentence2 = input("Enter Sentence 2: ")


# Generate sentence embeddings
vector1 = sentence_embedding(sentence1)
vector2 = sentence_embedding(sentence2)


# Check for unknown words
if vector1 is None or vector2 is None:

    print("\nError: Could not find words in GloVe vocabulary.")

else:

    # Calculate cosine similarity
    similarity = cosine_similarity(
        vector1.reshape(1, -1),
        vector2.reshape(1, -1)
    )[0][0]

    # Display result
    print("\n==============================================")
    print("          GLOVE SEMANTIC SIMILARITY")
    print("==============================================")

    print("Sentence 1 :", sentence1)
    print("Sentence 2 :", sentence2)

    print("----------------------------------------------")

    print("Cosine Similarity : {:.2f}".format(similarity))

    print("==============================================")

    if similarity >= 0.75:
        print("Result : Highly Similar")
    elif similarity >= 0.50:
        print("Result : Moderately Similar")
    elif similarity >= 0.25:
        print("Result : Slightly Similar")
    else:
        print("Result : Not Very Similar")

    print("==============================================")
```
