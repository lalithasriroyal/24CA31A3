```python
# =============================================================================
# Experiment 1: Text Cleaning - Removing Unwanted Characters and Normalizing Text
# =============================================================================
# Aim:
#     To perform text cleaning by removing unwanted characters and normalize
#     the text using Python.
#
# Objective:
#     Apply basic text cleaning techniques on a given text corpus using
#     Regular Expressions (Regex) and BeautifulSoup.
#
# Preprocessing Techniques Used:
#     1. Convert text to lowercase
#     2. Remove HTML tags
#     3. Remove URLs
#     4. Remove Email addresses
#     5. Remove Digits
#     6. Remove Punctuation and Special Characters
#     7. Remove Extra White Spaces
# =============================================================================


# =============================================================================
# Step 1: Create the Given Text Corpus
# =============================================================================

corpus = [
    "<p>Data Science is AMAZING!!! Visit https://datascience.com</p>",
    "Python 3.12 is released in 2025.",
    "NLP is used in AI.",
    "Contact us at info@gmail.com"
]


# =============================================================================
# Step 2: Import the Required Libraries
# =============================================================================

import re
from bs4 import BeautifulSoup


# =============================================================================
# Step 3: Convert All Text to Lowercase
# =============================================================================

lowercase_corpus = [
    text.lower()
    for text in corpus
]

print("=" * 70)
print("Step 3: Lowercase Conversion")
print("=" * 70)

for text in lowercase_corpus:
    print(text)


# =============================================================================
# Step 4: Remove HTML Tags
# =============================================================================

html_removed = [
    BeautifulSoup(text, "html.parser").get_text()
    for text in lowercase_corpus
]

print("\n" + "=" * 70)
print("Step 4: HTML Tags Removed")
print("=" * 70)

for text in html_removed:
    print(text)


# =============================================================================
# Step 5: Remove URLs
# =============================================================================

url_removed = [
    re.sub(r'https?://\S+|www\.\S+', '', text)
    for text in html_removed
]

print("\n" + "=" * 70)
print("Step 5: URLs Removed")
print("=" * 70)

for text in url_removed:
    print(text)


# =============================================================================
# Step 6: Remove Email Addresses
# =============================================================================

email_removed = [
    re.sub(r'\b[\w\.-]+@[\w\.-]+\.\w+\b', '', text)
    for text in url_removed
]

print("\n" + "=" * 70)
print("Step 6: Email Addresses Removed")
print("=" * 70)

for text in email_removed:
    print(text)


# =============================================================================
# Step 7: Remove Digits
# =============================================================================

digit_removed = [
    re.sub(r'\d+', '', text)
    for text in email_removed
]

print("\n" + "=" * 70)
print("Step 7: Digits Removed")
print("=" * 70)

for text in digit_removed:
    print(text)


# =============================================================================
# Step 8: Remove Punctuation and Special Characters
# =============================================================================

punct_removed = [
    re.sub(r'\W+', ' ', text)
    for text in digit_removed
]

print("\n" + "=" * 70)
print("Step 8: Punctuation and Special Characters Removed")
print("=" * 70)

for text in punct_removed:
    print(text)


# =============================================================================
# Step 9: Remove Extra White Spaces
# =============================================================================

cleaned_corpus = [
    re.sub(r'\s+', ' ', text).strip()
    for text in punct_removed
]

print("\n" + "=" * 70)
print("Step 9: Extra White Spaces Removed")
print("=" * 70)

for text in cleaned_corpus:
    print(text)


# =============================================================================
# Step 10: Store the Cleaned Text
# =============================================================================
# The cleaned text has already been stored in the list 'cleaned_corpus'.

print("\n" + "=" * 70)
print("Step 10: Cleaned Corpus Stored Successfully")
print("=" * 70)

print(cleaned_corpus)


# =============================================================================
# Step 11: Display the Original Corpus and Cleaned Corpus
# =============================================================================

print("\n" + "=" * 70)
print("Original Corpus")
print("=" * 70)

for sentence in corpus:
    print(sentence)

print("\n" + "=" * 70)
print("Cleaned Corpus")
print("=" * 70)

for sentence in cleaned_corpus:
    print(sentence)


# =============================================================================
# Step 12: Compare Before and After Cleaning
# =============================================================================

print("\n" + "=" * 70)
print("Comparison: Original Text vs Cleaned Text")
print("=" * 70)

for original, cleaned in zip(corpus, cleaned_corpus):

    print("Original :", original)
    print("Cleaned  :", cleaned)
    print("-" * 70)


# =============================================================================
# End of Program
# =============================================================================
```

### Live Scenario

ABC Electronics is a popular e-commerce company that sells smartphones, laptops, and electronic accessories. Every day, customers post reviews about their shopping experience on the company's website.

Before performing **Sentiment Analysis**, the Data Analytics team must clean the customer reviews because they contain HTML tags, URLs, email addresses, numbers, punctuation marks, emojis, and unnecessary white spaces.

#### Sample Dataset

```python
# Customer Reviews

reviews = [
    "<p>Excellent phone!!! 😍 Battery backup is amazing. Visit https://abcshop.com</p>",
    "Worst laptop ever!! 😡 I paid $999 in 2025. Contact support@abcshop.com",
    "Fast delivery 🚚 and excellent packaging. Highly Recommended!!!",
    "<div>Camera quality is AWESOME 📸. Worth every penny!!!</div>",
    "Average product 😐 Delivery was delayed by 3 days. Visit www.shippingupdates.com"
]

print("Original Reviews\n")

for review in reviews:
    print(review)
```

#### Import Libraries

```python
import re
from bs4 import BeautifulSoup
```

#### Clean the Reviews

```python
cleaned_reviews = []

for review in reviews:

    # Convert to lowercase
    review = review.lower()

    # Remove HTML tags
    review = BeautifulSoup(review, "html.parser").get_text()

    # Remove URLs
    review = re.sub(r'https?://\S+|www\.\S+', '', review)

    # Remove Email IDs
    review = re.sub(r'\b[\w\.-]+@[\w\.-]+\.\w+\b', '', review)

    # Remove digits
    review = re.sub(r'\d+', '', review)

    # Remove punctuation and special characters
    review = re.sub(r'\W+', ' ', review)

    # Remove extra spaces
    review = re.sub(r'\s+', ' ', review).strip()

    cleaned_reviews.append(review)
```

#### Compare Before and After

```python
print("Comparison of Original and Cleaned Reviews\n")

for original, cleaned in zip(reviews, cleaned_reviews):

    print("Original Review")
    print(original)

    print("\nCleaned Review")
    print(cleaned)

    print("-"*80)
```
