```python
import re
import numpy as np
from gensim.models import Word2Vec

print("Libraries imported successfully!")

# Input text corpus
corpus = [
    "Machine learning is a branch of artificial intelligence",
    "Machine learning uses algorithms to learn from data",
    "Artificial intelligence includes machine learning",
    "Deep learning is a part of machine learning",
   
    "Intelligence allows machines to solve problems",
   
    "Machine learning models analyze large data",
    
    "Deep learning uses neural networks",
    "Neural networks are used in machine learning",
    "Data science and machine learning work together",
    
    "Modern technology uses artificial intelligence",
   
    "Learning helps computers understand data",
    "Machine learning is useful in data science",
   
    "Deep learning helps machines recognize patterns"
]

print("\nInput Corpus:")
for i, sentence in enumerate(corpus, 1):
    print(i, ":", sentence)
```

```python
# Tokenization function
def tokenize(sentence):
    return re.findall(r'\b[a-zA-Z]+\b', sentence.lower())


# Tokenize the complete corpus
tokenized_corpus = [tokenize(sentence) for sentence in corpus]

print("TOKENIZED CORPUS")
print("=" * 60)

for i, sentence in enumerate(tokenized_corpus, 1):
    print(i, ":", sentence)
```

```python
# Train Word2Vec using CBOW
model = Word2Vec(
    sentences=tokenized_corpus,
    vector_size=50,
    window=3,
    min_count=1,
    sg=0,          # CBOW architecture
    workers=1,
    epochs=200,
    seed=42
)

print("Word2Vec CBOW model trained successfully!")
```

```python
print("VOCABULARY")
print("=" * 50)

vocabulary = model.wv.index_to_key

print("Total vocabulary words:", len(vocabulary))
print()

for i, word in enumerate(vocabulary, 1):
    print(i, ":", word)
```

```python
# Retrieve embedding vectors
machine_vector = model.wv["machine"]
learning_vector = model.wv["learning"]

print("EMBEDDING VECTORS")
print("=" * 60)

print("\nMachine Vector:")
print(machine_vector)

print("\nLearning Vector:")
print(learning_vector)

print("\nVector Dimension:")
print("Machine:", machine_vector.shape)
print("Learning:", learning_vector.shape)
```

```python
print("TOP 5 WORDS SIMILAR TO 'LEARNING'")
print("=" * 60)

similar_words = model.wv.most_similar("learning", topn=5)

for i, (word, score) in enumerate(similar_words, 1):
    print(i, ".", word, "->", round(score, 4))
```

```python
machine_learning_similarity = model.wv.similarity(
    "machine",
    "learning"
)

print("Machine - Learning Similarity:")
print(round(machine_learning_similarity, 4))
```

```python
artificial_intelligence_similarity = model.wv.similarity(
    "artificial",
    "intelligence"
)

print("Artificial - Intelligence Similarity:")
print(round(artificial_intelligence_similarity, 4))
```

```python
print("\n")
print("=" * 65)
print("           WORD2VEC CBOW SEMANTIC SIMILARITY")
print("=" * 65)

print("\n1. MODEL INFORMATION")
print("-" * 65)
print("Architecture        : CBOW")
print("sg                  :", 0)
print("Vector Dimension    :", model.vector_size)
print("Context Window      :", model.window)
print("Training Epochs     :", 200)
print("Vocabulary Size     :", len(model.wv))

print("\n2. WORD EMBEDDINGS")
print("-" * 65)

print("Machine vector dimension :", model.wv["machine"].shape)
print("Learning vector dimension:", model.wv["learning"].shape)

print("\n3. TOP 5 WORDS SIMILAR TO 'LEARNING'")
print("-" * 65)

for i, (word, score) in enumerate(
    model.wv.most_similar("learning", topn=5), 1
):
    print(f"{i}. {word:<20} Similarity = {score:.4f}")

print("\n4. SEMANTIC SIMILARITY SCORES")
print("-" * 65)

print(
    f"Machine - Learning       : "
    f"{model.wv.similarity('machine', 'learning'):.4f}"
)

print(
    f"Artificial - Intelligence: "
    f"{model.wv.similarity('artificial', 'intelligence'):.4f}"
)

print("\n5. INTERPRETATION")
print("-" * 65)

ml_score = model.wv.similarity("machine", "learning")
ai_score = model.wv.similarity("artificial", "intelligence")

if ml_score > 0:
    print("Machine and Learning have a positive semantic relationship.")

if ai_score > 0:
    print("Artificial and Intelligence have a positive semantic relationship.")

print("\nThe Word2Vec CBOW model successfully learned")
print("numerical word representations from the input corpus.")

print("=" * 65)
```
