```python
from transformers import pipeline

print("Libraries imported successfully!")

sentences = [
    "Sundar Pichai is the CEO of Google.",
    "Microsoft was founded by Bill Gates.",
    "The conference will be held in Chennai on 20 July 2026.",
    "Elon Musk visited India last week."
]

print("\nInput Sentences:")
for i, sentence in enumerate(sentences, 1):
    print(f"S{i}: {sentence}")
```

```python
# Load pretrained BERT NER model
ner_model = pipeline(
    "ner",
    model="dslim/bert-base-NER",
    aggregation_strategy="first"
)

print("Pretrained BERT NER model loaded successfully!")
```

```python
print("Named Entity Recognition Results")
print("=" * 60)

for i, sentence in enumerate(sentences, 1):

    entities = ner_model(sentence)

    print(f"\nS{i}: {sentence}")

    for entity in entities:
        print(
            f"Entity: {entity['word']:<20} "
            f"Type: {entity['entity_group']:<10} "
            f"Score: {entity['score']:.4f}"
        )
```

```python
print("\n" + "=" * 60)
print("             BERT NAMED ENTITY RECOGNITION")
print("=" * 60)

for i, sentence in enumerate(sentences, 1):

    entities = ner_model(sentence)

    print(f"\nS{i}: {sentence}")
    print("-" * 60)

    for entity in entities:
        print(
            f"{entity['word']}  -->  {entity['entity_group']}"
        )
```

```python
from transformers import pipeline

print("Libraries imported successfully!")

# Post-Lab sentences
sentences = [
    "Satya Nadella visited Hyderabad.",
    "Apple launched a new iPhone in California."
]

print("\nPost-Lab Input Sentences:")
for i, sentence in enumerate(sentences, 1):
    print(f"S{i}: {sentence}")
```

```python
print("Named Entity Recognition Results")
print("=" * 60)

for i, sentence in enumerate(sentences, 1):

    entities = ner_model(sentence)

    print(f"\nS{i}: {sentence}")

    for entity in entities:
        print(
            f"Entity: {entity['word']:<20} "
            f"Type: {entity['entity_group']:<10} "
            f"Score: {entity['score']:.4f}"
        )
```

```python
print("\n" + "=" * 60)
print("             POST-LAB BERT NER RESULTS")
print("=" * 60)

for i, sentence in enumerate(sentences, 1):

    entities = ner_model(sentence)

    print(f"\nS{i}: {sentence}")
    print("-" * 60)

    for entity in entities:
        print(f"{entity['word']} --> {entity['entity_group']}")

print("\n" + "=" * 60)
```
