### 3.Calculate the Term Frequency (TF) and  write formula ?
A. Term Frequency (TF) is the number of times a word appears in a document divided by the total number of words in that document.
Formula: TF = (Number of times the term appears) / (Total number of terms in the document).

### 4. Write formula and Calculate the Document Frequency (DF) ?
A.Document Frequency (DF) is the number of documents in which a particular word appears at least once.
Formula: DF(term) = Number of documents containing the term.

### 5.Write formula and fine Inverse Document Frequency (IDF)? 
A.Inverse Document Frequency (IDF) measures how important or rare a word is across all documents.
Formula: IDF = log(Total number of documents / Number of documents containing the term).

### 6.Write the formula for calculating TF-IDF and give the table?
A.Formula: TF-IDF = TF × IDF.
Table: | Word | TF | IDF | TF-IDF |.

### Activity 1: Vocabulary Creation 

Tasks 
1. Import the required Python libraries.
2. Create the input corpus.
3. Generate the vocabulary.
4. Display the vocabulary.  

```python
# Import library
import math
import pandas as pd

# Create input corpus
documents = [
    "I love NLP",
    "I love Python",
    "NLP is fun"
]

# Generate vocabulary
vocabulary = set()

for doc in documents:
    words = doc.lower().split()
    vocabulary.update(words)

vocabulary = sorted(vocabulary)

# Display vocabulary
print("Vocabulary:")
print(vocabulary)
print("Vocabulary Size:", len(vocabulary))
```

### Activity 2: Term Frequency (TF), Document Frequency (DF) and Inverse Document Frequency (IDF), TF-IDF Matrix 
Tasks :
1. Calculate the total number of words in each document.  
2. Compute the Term Frequency (TF) of each word.  
3. Display the TF matrix. 
4. Calculate the Document Frequency (DF) of each term.  
5. Compute the Inverse Document Frequency (IDF). 
6. Display the DF and IDF values in tabular format. Display the TF-IDF matrix.  


```python
# Import libraries
import math
import pandas as pd

# Input corpus
documents = [
    "I love NLP",
    "I love Python",
    "NLP is fun"
]

# Create vocabulary
vocabulary = sorted(set(word for doc in documents
                        for word in doc.lower().split()))

# Calculate total words in each document
total_words = [len(doc.lower().split()) for doc in documents]

# Calculate TF matrix
tf_matrix = []

for doc in documents:
    words = doc.lower().split()
    tf_row = []

    for word in vocabulary:
        tf = words.count(word) / len(words)
        tf_row.append(tf)

    tf_matrix.append(tf_row)

tf_df = pd.DataFrame(tf_matrix, columns=vocabulary,
                     index=["D1", "D2", "D3"])

print("Total words in each document:")
print(total_words)

print("\nTF Matrix:")
print(tf_df)

# Calculate DF
df_values = []

for word in vocabulary:
    df = sum(word in doc.lower().split() for doc in documents)
    df_values.append(df)

# Calculate IDF
N = len(documents)
idf_values = [math.log10(N / df) for df in df_values]

# Display DF and IDF
df_idf = pd.DataFrame({
    "Term": vocabulary,
    "DF": df_values,
    "IDF": idf_values
})

print("\nDF and IDF:")
print(df_idf)

# Calculate TF-IDF
tfidf_matrix = []

for i in range(len(documents)):
    row = []
    for j in range(len(vocabulary)):
        tfidf = tf_matrix[i][j] * idf_values[j]
        row.append(tfidf)
    tfidf_matrix.append(row)

tfidf_df = pd.DataFrame(tfidf_matrix,
                        columns=vocabulary,
                        index=["D1", "D2", "D3"])

print("\nTF-IDF Matrix:")
print(tfidf_df)
```
